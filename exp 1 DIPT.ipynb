```python
import cv2
import matplotlib.pyplot as plt
pic = cv2.imread("photo.png")
img = cv2.imread("dog.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
pic_rgb = cv2.cvtColor(pic, cv2.COLOR_BGR2RGB)
pic_re = cv2.resize(pic_rgb,(200,200))
img_rgb[225:425, 120:320] = pic_re
plt.imshow(img_rgb)
plt.axis("off")
plt.title("IMAGE WITH 200x200 WHITE BLOCK")
plt.show()
```

```python
img = cv2.imread('dog.jpg')
flip_hor = cv2.flip(image,1)
flip_hor_rgb = cv2.cvtColor(flip_hor,cv2.COLOR_BGR2RGB)
plt.imshow(flip_hor_rgb)
plt.axis('off')
plt.title("FLIPPED HORIZONTALLY")
plt.show()
```
